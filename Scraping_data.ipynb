```python
from google_play_scraper import app, reviews, Sort, reviews_all
import pandas as pd
```

```python
from google_play_scraper import reviews, Sort

scrape_review, _ = reviews(
    'com.lemon.lvoverseas',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=11000
)
```

```python
import csv

with open('review_capcut.csv', mode='w', newline='', encoding='utf-8') as file :
  writer = csv.writer(file)
  writer.writerow(['Review'])
  for review in scrape_review :
    writer.writerow([review['content']])
```

```python
reviews_capcut_df = pd.DataFrame(scrape_review)
reviews_capcut_df.shape
reviews_capcut_df.head()
reviews_capcut_df.to_csv('review_capcut.csv', index=False)

reviews_capcut_df = pd.DataFrame(scrape_review)

jumlah_review, jumlah_kolom = reviews_capcut_df.shape
print(f"Jumlah review: {jumlah_review}")
print(f"Jumlah kolom: {jumlah_kolom}")

reviews_capcut_df.head()
```
